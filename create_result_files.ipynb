```python
import os
```

# Create new strategy result files for experiment

```python
def create_new_strategy_files(experiment_name: str, existing_strategy: str, new_strategy: str):
    for database in os.listdir("data"):
        database_path = os.path.join("data", database)

        if not os.path.isdir(database_path):
            continue

        for experiment in os.listdir(database_path):
            experiment_path = os.path.join(database_path, experiment)

            if experiment != experiment_name:
                continue

            if not os.path.isdir(experiment_path):
                continue

            os.makedirs(os.path.join(experiment_path, new_strategy), exist_ok=True)

            for file in os.listdir(os.path.join(experiment_path, existing_strategy)):
                open(
                    os.path.join(experiment_path, new_strategy, file),
                    "w",
                    encoding="utf-8",
                ).close()
```

```python
experiment_name = "paper_evaluation"
existing_strategy = "justine_optimization_2"
new_strategy = "Llama3_3_one_prompt"

create_new_strategy_files(experiment_name, existing_strategy, new_strategy)
```
